```python
import numpy as np

# Dataset - XOR problem (classic test for NNs)
X = np.array([[0,0], [0,1], [1,0], [1,1]])  # (4, 2)
y = np.array([[0], [1], [1], [0]])            # (4, 1)

# Initialize weights randomly, biases as zeros
np.random.seed(42)
W1 = np.random.randn(3, 2) * 0.01  # hidden layer (3 neurons, 2 inputs)
b1 = np.zeros((3, 1))
W2 = np.random.randn(1, 3) * 0.01  # output layer (1 neuron, 3 inputs)
b2 = np.zeros((1, 1))
```

```python
def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def forward(X):
    Z1 = W1 @ X.T + b1   # (3,2) @ (2,4) = (3,4)
    A1 = relu(Z1)          # (3,4)
    Z2 = W2 @ A1 + b2     # (1,3) @ (3,4) = (1,4)
    A2 = sigmoid(Z2)       # (1,4)
    return Z1, A1, Z2, A2

Z1, A1, Z2, A2 = forward(X)
print(A2)
```

```python
def compute_loss(A2, y):
    L = -np.mean(y.T * np.log(A2) + (1-y.T) * np.log(1-A2))
    return L

loss = compute_loss(A2, y)
print(loss)
```
